## Module 1: URL Validation

This module validates the website URL entered by the user.

It checks:
1. Empty URL input.
2. HTTP and HTTPS protocols.
3. Valid website address.
4. Invalid URL handling.

```python
# Module 1: URL Validation

from urllib.parse import urlparse

def validate_url(url):

    # Remove unnecessary spaces
    url = url.strip()

    # Check empty input
    if not url:
        return False, "URL cannot be empty."

    # Add HTTPS if protocol is missing
    if "://" not in url:
        url = "https://" + url

    # Parse the URL
    parsed_url = urlparse(url)

    # Check protocol
    if parsed_url.scheme not in ["http", "https"]:
        return False, "Only HTTP and HTTPS URLs are allowed."

    # Check website hostname
    if not parsed_url.hostname:
        return False, "Invalid website address."

    # Return valid URL
    return True, url


# Test the URL validation module

test_url = "https://example.com"

is_valid, result = validate_url(test_url)

print("SECURESCAN - URL VALIDATION")
print("-" * 35)

if is_valid:
    print("Status:", "Valid URL")
    print("Website:", result)
else:
    print("Status:", "Invalid URL")
    print("Reason:", result)
```

## Module 2: Website Connection and HTTPS Checker

This module connects to the target website and checks:

1. Website availability.
2. HTTP response status code.
3. HTTPS usage.
4. Response time.
5. Final website URL after redirection.

```python
# Module 2: Website Connection and HTTPS Checker

import requests
import time

def check_website(url):

    print("\nSECURESCAN - WEBSITE CONNECTION CHECK")
    print("-" * 45)

    try:
        start_time = time.time()

        response = requests.get(
            url,
            timeout=10
        )

        end_time = time.time()

        # Calculate response time
        response_time = round(
            (end_time - start_time) * 1000, 2
        )

        # Display website information
        print("Website:", url)
        print("Status Code:", response.status_code)
        print("Final URL:", response.url)
        print("Response Time:", response_time, "ms")

        # Check HTTPS
        if response.url.startswith("https://"):
            print("HTTPS Status: Enabled")
        else:
            print("HTTPS Status: Not detected")

        # Check website availability
        if response.status_code == 200:
            print("Website Status: Reachable")
        else:
            print("Website Status: Responded with status",
                  response.status_code)

    except requests.exceptions.Timeout:
        print("Error: Website connection timed out.")

    except requests.exceptions.RequestException as error:
        print("Connection Error:", error)


# Test using a public example website
check_website("https://example.com")
```

## Module 3: Security Headers Analyzer

This module analyzes the security headers returned by a website.

It checks:

1. Content-Security-Policy
2. Strict-Transport-Security
3. X-Content-Type-Options
4. X-Frame-Options
5. Referrer-Policy
6. Permissions-Policy

The module identifies which security headers are present
and which are missing.

```python
# Module 3: Security Headers Analyzer

import requests

def analyze_security_headers(url):

    print("\nSECURESCAN - SECURITY HEADERS ANALYSIS")
    print("=" * 50)

    # List of important security headers
    security_headers = {
        "Content-Security-Policy":
            "Helps control which content a browser can load.",

        "Strict-Transport-Security":
            "Instructs browsers to use HTTPS.",

        "X-Content-Type-Options":
            "Helps prevent MIME type sniffing.",

        "X-Frame-Options":
            "Helps protect against clickjacking.",

        "Referrer-Policy":
            "Controls referrer information shared by browsers.",

        "Permissions-Policy":
            "Controls access to browser features."
    }

    try:
        # Send request to the website
        response = requests.get(url, timeout=10)

        headers = response.headers

        present = 0
        missing = 0

        print("Website:", response.url)
        print("-" * 50)

        # Check each security header
        for header, description in security_headers.items():

            if header in headers:
                print("[PRESENT]", header)
                present += 1

            else:
                print("[MISSING]", header)
                missing += 1

        print("-" * 50)
        print("Total Headers Checked:", len(security_headers))
        print("Present Headers:", present)
        print("Missing Headers:", missing)

        print("=" * 50)

    except requests.exceptions.RequestException as error:
        print("Website connection failed:", error)


# Run Security Headers Analyzer
analyze_security_headers("https://example.com")
```

## Module 4: CORS Configuration Checker

This module checks the CORS-related HTTP response headers.

It analyzes:

1. Access-Control-Allow-Origin
2. Access-Control-Allow-Credentials
3. Access-Control-Allow-Methods
4. Access-Control-Allow-Headers

It identifies potentially risky configurations for further review.

```python
# Module 4: CORS Configuration Checker

import requests

def check_cors(url):

    print("\nSECURESCAN - CORS CONFIGURATION CHECK")
    print("=" * 50)

    try:
        response = requests.get(url, timeout=10)

        headers = response.headers

        # Read CORS headers
        allow_origin = headers.get(
            "Access-Control-Allow-Origin"
        )

        allow_credentials = headers.get(
            "Access-Control-Allow-Credentials"
        )

        allow_methods = headers.get(
            "Access-Control-Allow-Methods"
        )

        allow_headers = headers.get(
            "Access-Control-Allow-Headers"
        )

        print("Website:", response.url)
        print("-" * 50)

        print("Allowed Origin:", allow_origin or "Not specified")
        print("Credentials:", allow_credentials or "Not specified")
        print("Allowed Methods:", allow_methods or "Not specified")
        print("Allowed Headers:", allow_headers or "Not specified")

        print("-" * 50)

        # Basic configuration observations
        if allow_origin == "*":
            print("Observation: Wildcard origin is allowed.")

        if (
            allow_origin == "*"
            and str(allow_credentials).lower() == "true"
        ):
            print("Warning: Review wildcard origin and credentials configuration.")

        if not allow_origin:
            print("Observation: No CORS allow-origin header found.")

        print("=" * 50)

    except requests.exceptions.RequestException as error:
        print("CORS check failed:", error)


# Run CORS Checker
check_cors("https://example.com")
```

## Module 5: Security Findings Analysis

This module combines important website security observations, including HTTPS status, security headers, CORS configuration, HTTP response status, and response time.

The scanner displays findings as PASS, REVIEW, or INFO. These findings are basic observations and do not confirm that a website is completely secure or vulnerable.

```python
import requests
import time
from urllib.parse import urlparse

def analyze_security_findings(url):

    is_valid, validated_url = validate_url(url)

    if not is_valid:
        print("Invalid URL:", validated_url)
        return None

    print("\nSECURESCAN - SECURITY FINDINGS ANALYSIS")
    print("=" * 55)

    try:
        start_time = time.time()

        response = requests.get(
            validated_url,
            timeout=10,
            allow_redirects=True
        )

        response_time = round(
            (time.time() - start_time) * 1000, 2
        )

        final_url = response.url
        headers = response.headers
        findings = []

        # 1. HTTPS Check
        if urlparse(final_url).scheme == "https":
            findings.append(
                ("HTTPS", "PASS", "Website uses HTTPS.")
            )
        else:
            findings.append(
                ("HTTPS", "REVIEW", "Final URL does not use HTTPS.")
            )

        # 2. Security Headers Check
        required_headers = [
            "Content-Security-Policy",
            "Strict-Transport-Security",
            "X-Content-Type-Options",
            "X-Frame-Options",
            "Referrer-Policy",
            "Permissions-Policy"
        ]

        missing_headers = [
            header for header in required_headers
            if header not in headers
        ]

        if missing_headers:
            findings.append(
                (
                    "Security Headers",
                    "REVIEW",
                    "Missing: " + ", ".join(missing_headers)
                )
            )
        else:
            findings.append(
                (
                    "Security Headers",
                    "PASS",
                    "All selected headers are present."
                )
            )

        # 3. CORS Check
        allow_origin = headers.get(
            "Access-Control-Allow-Origin"
        )
        allow_credentials = headers.get(
            "Access-Control-Allow-Credentials", ""
        )

        if allow_origin == "*" and allow_credentials.lower() == "true":
            findings.append(
                (
                    "CORS",
                    "REVIEW",
                    "Wildcard origin and credentials settings need review."
                )
            )
        elif allow_origin == "*":
            findings.append(
                (
                    "CORS",
                    "INFO",
                    "Wildcard origin is configured."
                )
            )
        else:
            findings.append(
                (
                    "CORS",
                    "INFO",
                    "Allow-Origin: " + str(allow_origin or "Not specified")
                )
            )

        # 4. HTTP Response Check
        findings.append(
            (
                "HTTP Response",
                "INFO",
                "Server returned status code " + str(response.status_code)
            )
        )

        # 5. Response Time
        findings.append(
            (
                "Response Time",
                "INFO",
                str(response_time) + " ms"
            )
        )

        # Display Results
        print("Website:", final_url)
        print("-" * 55)

        for check, status, details in findings:
            print(f"[{status}] {check}")
            print("Details:", details)
            print("-" * 55)

        print("Analysis completed successfully.")
        print("=" * 55)

        return {
            "website": final_url,
            "status_code": response.status_code,
            "response_time_ms": response_time,
            "findings": findings
        }

    except requests.exceptions.RequestException as error:
        print("Scan failed:", error)
        return None


scan_results = analyze_security_findings("https://example.com")
```

## Module 6: Save Scan Results

This module saves the website security scan results in a JSON file.

The saved file contains the scanned website URL, HTTP status code, response time, and security findings.

```python
import json
import os
from datetime import datetime

def save_scan_results(results):

    if results is None:
        print("No scan results available to save.")
        return

    os.makedirs("reports", exist_ok=True)

    results["scan_date"] = datetime.now().strftime(
        "%Y-%m-%d %H:%M:%S"
    )

    filename = "reports/SecureScan_Report.json"

    try:
        with open(filename, "w", encoding="utf-8") as file:
            json.dump(results, file, indent=4)

        print("\nSECURESCAN - REPORT SAVED")
        print("=" * 45)
        print("File Name:", filename)
        print("Status: Successfully Saved")
        print("=" * 45)

    except (OSError, TypeError) as error:
        print("Error saving report:", error)


save_scan_results(scan_results)
```

## Module 7: HTML Security Report Generation

This module generates a formatted HTML report using the saved website security scan results.

The report displays website details, scan date, HTTP status, response time, and security findings.

```python
import os
from html import escape

def generate_html_report(results):

    if results is None:
        print("No scan results available.")
        return

    os.makedirs("reports", exist_ok=True)

    website = escape(str(results.get("website", "N/A")))
    scan_date = escape(str(results.get("scan_date", "N/A")))
    status_code = escape(str(results.get("status_code", "N/A")))
    response_time = escape(
        str(results.get("response_time_ms", "N/A"))
    )

    rows = ""

    for check, status, details in results.get("findings", []):
        rows += f"""
        <tr>
            <td>{escape(str(check))}</td>
            <td>{escape(str(status))}</td>
            <td>{escape(str(details))}</td>
        </tr>
        """

    html_content = f"""
    <!DOCTYPE html>
    <html lang="en">
    <head>
        <meta charset="UTF-8">
        <meta name="viewport" content="width=device-width, initial-scale=1.0">
        <title>SecureScan Security Report</title>

        <style>
            body {{
                font-family: Arial, sans-serif;
                background: #f3f4f8;
                margin: 0;
                padding: 30px;
                color: #222;
            }}

            .container {{
                max-width: 900px;
                margin: auto;
                background: white;
                padding: 30px;
                border-radius: 12px;
            }}

            h1 {{
                color: #4932a8;
                text-align: center;
            }}

            .details {{
                background: #f0edff;
                padding: 15px;
                border-radius: 8px;
                line-height: 1.8;
            }}

            table {{
                width: 100%;
                border-collapse: collapse;
                margin-top: 20px;
            }}

            th, td {{
                border: 1px solid #ddd;
                padding: 12px;
                text-align: left;
                overflow-wrap: anywhere;
            }}

            th {{
                background: #4932a8;
                color: white;
            }}

            tr:nth-child(even) {{
                background: #f7f7fb;
            }}

            .note {{
                margin-top: 25px;
                color: #666;
                font-size: 13px;
            }}
        </style>
    </head>

    <body>
        <div class="container">

            <h1>SECURESCAN</h1>
            <h2>Website Security Assessment Report</h2>

            <div class="details">
                <b>Website:</b> {website}<br>
                <b>Scan Date:</b> {scan_date}<br>
                <b>HTTP Status Code:</b> {status_code}<br>
                <b>Response Time:</b> {response_time} ms
            </div>

            <h2>Security Findings</h2>

            <table>
                <tr>
                    <th>Security Check</th>
                    <th>Status</th>
                    <th>Details</th>
                </tr>

                {rows}

            </table>

            <p class="note">
                Disclaimer: This report contains basic security observations.
                It is not a complete penetration test or guarantee of website security.
            </p>

            <p class="note">
                Developed by: Chetana S. Tisage
            </p>

        </div>
    </body>
    </html>
    """

    filename = "reports/SecureScan_Report.html"

    try:
        with open(filename, "w", encoding="utf-8") as file:
            file.write(html_content)

        print("HTML report generated successfully!")
        print("Saved at:", filename)

    except OSError as error:
        print("Error generating HTML report:", error)


generate_html_report(scan_results)
```
